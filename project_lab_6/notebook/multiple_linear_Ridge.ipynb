```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.linear_model import (
    LinearRegression,
    Ridge,
    RidgeCV,
    Lasso,
    LassoCV
)

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
data = {
    'Hours_studied': [2, 4, 6, 7, 8],
    'Classes_Attended': [3, 4, 5, 6, 8],
    'Final_score': [65, 70, 75, 85, 95]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)
```

```python
# For Multiple Linear Regression
X = df[['Hours_studied', 'Classes_Attended']]

# Target variable
y = df['Final_score']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,
    random_state=42
)

print("\nTraining data:")
print(X_train)

print("\nTesting data:")
print(X_test)
```

 SIMPLE LINEAR REGRESSION

```python
# Use only Hours_studied
X_simple = df[['Hours_studied']]

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_simple,
    y,
    test_size=0.4,
    random_state=42
)
```

```python

simple_model = LinearRegression()

simple_model.fit(X_train_s, y_train_s)

y_pred_simple = simple_model.predict(X_test_s)
```

```python
print("\n========================================")
print("ORDINARY / SIMPLE LINEAR REGRESSION")
print("========================================")

print("Intercept:", simple_model.intercept_)
print("Coefficient:", simple_model.coef_)

print("Actual:", y_test_s.values)
print("Predicted:", y_pred_simple)
```

MULTIPLE LINEAR REGRESSION

```python
multiple_model = LinearRegression()

multiple_model.fit(X_train, y_train)

y_pred_multiple = multiple_model.predict(X_test)
```

```python
print("\n========================================")
print("MULTIPLE LINEAR REGRESSION")
print("========================================")

print("Intercept:", multiple_model.intercept_)
print("Coefficients:", multiple_model.coef_)

print("Actual:", y_test.values)
print("Predicted:", y_pred_multiple)
```

 RIDGE REGRESSION

```python
ridge_model = make_pipeline(
    StandardScaler(),
    Ridge(alpha=1.0)
)

ridge_model.fit(X_train, y_train)

y_pred_ridge = ridge_model.predict(X_test)
```

```python
print("\n========================================")
print("RIDGE REGRESSION")
print("========================================")

print("Alpha:", 1.0)

print("Actual:", y_test.values)
print("Predicted:", y_pred_ridge)
```

RIDGE CV

```python
ridge_cv_model = make_pipeline(
    StandardScaler(),
    RidgeCV(
        alphas=[0.01, 0.1, 1, 10, 100],
        cv=3
    )
)

```

```python
ridge_cv_model.fit(X_train, y_train)

y_pred_ridge_cv = ridge_cv_model.predict(X_test)

best_ridge_alpha = ridge_cv_model.named_steps['ridgecv'].alpha_
```

```python
print("\n========================================")
print("RIDGE CV")
print("========================================")

print("Best Alpha:", best_ridge_alpha)

print("Actual:", y_test.values)
print("Predicted:", y_pred_ridge_cv)
```

LASSO REGRESSION

```python
lasso_model = make_pipeline(
    StandardScaler(),
    Lasso(alpha=0.1)
)

lasso_model.fit(X_train, y_train)

y_pred_lasso = lasso_model.predict(X_test)
```

```python
print("\n========================================")
print("LASSO REGRESSION")
print("========================================")

print("Alpha:", 0.1)

print("Actual:", y_test.values)
print("Predicted:", y_pred_lasso)
```

LASSO CV

```python

lasso_cv_model = make_pipeline(
    StandardScaler(),
    LassoCV(
        alphas=[0.01, 0.1, 1, 10, 100],
        cv=3,
        max_iter=10000
    )
)

lasso_cv_model.fit(X_train, y_train)

y_pred_lasso_cv = lasso_cv_model.predict(X_test)

best_lasso_alpha = lasso_cv_model.named_steps['lassocv'].alpha_
```

```python
print("\n========================================")
print("LASSO CV")
print("========================================")

print("Best Alpha:", best_lasso_alpha)

print("Actual:", y_test.values)
print("Predicted:", y_pred_lasso_cv)
```

FUNCTION TO CALCULATE ERRORS

```python
def calculate_metrics(y_actual, y_pred):

    mae = mean_absolute_error(y_actual, y_pred)

    mse = mean_squared_error(y_actual, y_pred)

    r2 = r2_score(y_actual, y_pred)

    return mae, mse, r2
```

CALCULATE METRICS FOR ALL MODELS

```python
#for simple regression
mae_simple, mse_simple, r2_simple = calculate_metrics(
    y_test_s,
    y_pred_simple
)
print("MAE:", mae_simple)
print("MSE:", mse_simple)
print("R2 Score:", r2_simple)
```

```python
#for multiple regression
mae_multiple, mse_multiple, r2_multiple = calculate_metrics(
    y_test,
    y_pred_multiple
)
print("MAE:", mae_multiple)
print("MSE:", mse_multiple)
print("R2 Score:", r2_multiple)
```

```python
#Ridge Regression
mae_ridge, mse_ridge, r2_ridge = calculate_metrics(
    y_test,
    y_pred_ridge
)

print("MAE:", mae_ridge)
print("MSE:", mse_ridge)
print("R2 Score:", r2_ridge)
```

```python
# for RIDGE CV
mae_ridge_cv, mse_ridge_cv, r2_ridge_cv = calculate_metrics(
    y_test,
    y_pred_ridge_cv
)

print("MAE:", mae_ridge_cv)
print("MSE:", mse_ridge_cv)
print("R2 Score:", r2_ridge_cv)
print("Best Alpha:", best_ridge_alpha)
```

```python
# for LASSO REGRESSION
mae_lasso, mse_lasso, r2_lasso = calculate_metrics(
    y_test,
    y_pred_lasso
)

print("MAE:", mae_lasso)
print("MSE:", mse_lasso)
print("R2 Score:", r2_lasso)
```

```python
#for /LASSO CV
mae_lasso_cv, mse_lasso_cv, r2_lasso_cv = calculate_metrics(
    y_test,
    y_pred_lasso_cv
)

print("MAE:", mae_lasso_cv)
print("MSE:", mse_lasso_cv)
print("R2 Score:", r2_lasso_cv)
print("Best Alpha:", best_lasso_alpha)
```
