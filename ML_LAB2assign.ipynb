9. Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

Using NumPy boolean indexing, find:

Marks greater than 75
Marks between 50 and 80
Number of students who scored below 50
Percentage of students who scored 60 or above

Do not use a Python for loop.

```python
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
```

```python
print("Marks greater than 75:", marks[marks > 75])

print("Marks between 50 and 80:",
      marks[(marks >= 50) & (marks <= 80)])

print("Number of students who scored below 50:",
      np.sum(marks < 50))

print("Percentage of students who scored 60 or above:",
      np.mean(marks >= 60) * 100, "%")
```

10. The following array contains marks of four students in three subjects:

marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

Calculate:

Total marks of each student
Average marks of each student
Average marks in each subject
Student with the highest total
Highest mark in each subject.

```python
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])
```

```python

# Total marks of each student
student_totals = marks.sum(axis=1)

# Average marks of each student
student_averages = marks.mean(axis=1)

# Average marks in each subject
subject_averages = marks.mean(axis=0)

# Student with highest total
highest_student = np.argmax(student_totals) + 1

# Highest mark in each subject
highest_subject_marks = marks.max(axis=0)

print("Total marks of each student:", student_totals)
print("Average marks of each student:", student_averages)
print("Average marks in each subject:", subject_averages)
print("Student with highest total: Student", highest_student)
print("Highest mark in each subject:", highest_subject_marks)
```

12. Daily temperatures for two weeks are:

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

Find:

Mean temperature
Maximum and minimum temperatures
Number of days above the average temperature
Temperatures greater than 34°C
Difference between maximum and minimum
Standard deviation

```python
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])
```

```python
mean_temp = temperature.mean()
max_temp = temperature.max()
min_temp = temperature.min()
above_average = np.sum(temperature > mean_temp)
greater_34 = temperature[temperature > 34]
difference = max_temp - min_temp
std = temperature.std()
```

```python
print("Mean temperature:", mean_temp)
print("Maximum temperature:", max_temp)
print("Minimum temperature:", min_temp)
print("Days above average:", above_average)
print("Temperatures greater than 34°C:", greater_34)
print("Difference:", difference)
print("Standard deviation:", std)
```

13. Monthly sales of a company are:
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

Determine:

Total annual sales
Average monthly sales
Month with the highest sales
Month with the lowest sales
Number of months with sales above ₹20,000
Percentage increase from January to December

Hint: Remember that NumPy indexing starts at 0, while month numbering normally starts at 1.

```python
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])
```

```python
average_sales = sales.mean()
print(average_sales)
```

14. Consider marks of five students in four subjects:

marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

Write a NumPy program that calculates:

Total marks of every student
Average of every student
Average mark for every subject
Highest-scoring student
Lowest-scoring student
Students whose average is greater than 80
Overall class average

Then assign a performance category based on average:

|  Average | Category          |
| -------: | ----------------- |
|     ≥ 85 | Excellent         |
| 70–84.99 | Good              |
| 50–69.99 | Average           |
|     < 50 | Needs Improvement |

Challenge: Try assigning these categories using np.where() or np.select() instead of a Python loop.

```python
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])
```

```python
# Total marks
total = marks.sum(axis=1)

print("Total marks:", total)
```

```python
# Average marks of every student
average = marks.mean(axis=1)
print("Average marks:", average)

```

```python
# Average mark for every subject
subject_average = marks.mean(axis=0)
print("Subject averages:", subject_average)

```

```python
# Highest-scoring student
highest_student = np.argmax(total) + 1

print("Highest-scoring student:", highest_student)
```

```python
# Lowest-scoring student
lowest_student = np.argmin(total) + 1

print("Lowest-scoring student:", lowest_student)
```

```python
# Students with average > 80
students_above_80 = np.where(average > 80)[0] + 1

print("Students with average > 80:", students_above_80)
```

```python
# Overall class average
class_average = marks.mean()

print("Overall class average:", class_average)
```

```python
# Performance categories
categories = np.select(
    [
        average >= 85,
        average >= 70,
        average >= 50
    ],
    [
        "Excellent",
        "Good",
        "Average"
    ],
    default="Needs Improvement"
)
print("Performance categories:", categories)
```

23. Create a NumPy array containing 100 random exam marks and determine how many students fall into the ranges 0–39, 40–59, 60–79, and 80–100.

```python
marks = np.random.randint(0, 101, 100)
marks
```

```python
range_0_39 = np.sum((marks >= 0) & (marks <= 39))
range_40_59 = np.sum((marks >= 40) & (marks <= 59))
range_60_79 = np.sum((marks >= 60) & (marks <= 79))
range_80_100 = np.sum((marks >= 80) & (marks <= 100))

print("Marks:", marks)

print("0-39:", range_0_39)
print("40-59:", range_40_59)
print("60-79:", range_60_79)
print("80-100:", range_80_100)

print(
    "Total:",
    range_0_39 + range_40_59 +
    range_60_79 + range_80_100
)
```

24. Compare the time required to multiply 1,000,000 numbers using a Python list and a NumPy array.

```python
import time

n = 1000000

# Python list
python_list = list(range(n))

start = time.time()

list_result = [x * 2 for x in python_list]

list_time = time.time() - start


# NumPy array
numpy_array = np.arange(n)

start = time.time()

numpy_result = numpy_array * 2

numpy_time = time.time() - start


print("Python list time:", list_time, "seconds")
print("NumPy array time:", numpy_time, "seconds")
```
