```python
#matrix creation by array input
num=np.matrix(([11,121,130],[613,30,20],[589,22,81]))
```

```python
type(num)
```

```python
print("Main diagonal elements : \n",np.diag(num))
print("\n Daigonal elements above main diagonal : \n",np.diag(num,1))
print("\n Daigonal elements below main diagonal : \n",np.diag(num,-1))
print("\n Daigonal elements one step above main diagonal : \n",np.diag(num,2))
print("\n Daigonal elements one step below main diagonal : \n",np.diag(num,-2))
```

```python
x=np.array([1,2,3])
y=np.array([4,5,6])
print(x)
print(y)
```

```python
print(x+y) #element wise addition
print(x-y) #substraction
print(x*y) #multiplication
print(x/y) #division
```

```python
print(x**2) #element wise power
print(x@y) #dot product 1*4 + 2*5 +3*6 = 32
```

```python
#dot product:
x.dot(y)
```
