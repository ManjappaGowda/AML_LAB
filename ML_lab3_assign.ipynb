Find:

```python
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)
df
```

20. Create:

employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})


Determine:

Average salary.
Maximum and minimum salary.
Employee receiving the highest salary.
Employees earning above ₹60,000.
Average salary for each department.
Average experience by department.
Number of employees in each department.
Sort employees by salary.
Find IT employees with more than 3 years of experience.
Add a SalaryLevel column:
	≥ ₹75,000 → High
	₹50,000–₹74,999 → Medium
	< ₹50,000 → Low

```python
employees = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})

employees
```

```python
# 1. Average salary
print("1. Average salary:")
print(employees["Salary"].mean())
```

```python
# 2. Maximum and minimum salary
print("\n2. Maximum salary:")
print(employees["Salary"].max())

print("\nMinimum salary:")
print(employees["Salary"].min())
```

```python
# 3. Employee receiving the highest salary
print("\n3. Employee receiving highest salary:")
print(employees.loc[employees["Salary"].idxmax()])
```

```python
# 4. Employees earning above ₹60,000
print("\n4. Employees earning above ₹60,000:")
print(employees[employees["Salary"] > 60000])
```

```python
# 5. Average salary for each department
print("\n5. Average salary by department:")
print(employees.groupby("Department")["Salary"].mean())
```

```python
# 6. Average experience by department
print("\n6. Average experience by department:")
print(employees.groupby("Department")["Experience"].mean())
```

```python
# 7. Number of employees in each department
print("\n7. Number of employees in each department:")
print(employees.groupby("Department")["Employee"].count())

```

```python
# 8. Sort employees by salary
print("\n8. Employees sorted by salary:")
print(employees.sort_values("Salary", ascending=False))
```

```python
# 9. IT employees with more than 3 years of experience
print("\n9. IT employees with more than 3 years of experience:")
print(employees[
    (employees["Department"] == "IT") &
    (employees["Experience"] > 3)
])
```

```python
# 10. Add SalaryLevel column
employees["SalaryLevel"] = employees["Salary"].apply(
    lambda x: "High" if x >= 75000 else
              "Medium" if x >= 50000 else
              "Low"
)

print("\n10. DataFrame with SalaryLevel:")
print(employees)
```
