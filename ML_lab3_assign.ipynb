7. Find:

	Students scoring above 80.
	Students scoring between 75 and 90.
	Students belonging to CSE.
	CSE students scoring above 80.

```python
# Students scoring above 80
print("Students scoring above 80:")
print(df[df["Marks"] > 80])
```

```python
# Students scoring between 75 and 90
print("\nStudents scoring between 75 and 90:")
print(df[(df["Marks"] >= 75) & (df["Marks"] <= 90)])

```

```python
# Students belonging to CSE
print("\nStudents belonging to CSE:")
print(df[df["Department"] == "CSE"])
```

```python
# CSE students scoring above 80
print("\nCSE students scoring above 80:")
print(df[(df["Department"] == "CSE") & (df["Marks"] > 80)])
```

10. Assign grades according to:

	| Marks | Grade |
	| ----: | ----- |
	|  ≥ 90 | A     |
	|  ≥ 80 | B     |
	|  ≥ 70 | C     |
	|  < 70 | D     |

```python
# Assign grades
df["Grade"] = df["Marks"].apply(
    lambda x: "A" if x >= 90 else
              "B" if x >= 80 else
              "C" if x >= 70 else
              "D"
)

df
```

11. Sort students:
